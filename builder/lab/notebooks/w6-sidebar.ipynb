## The idea

You hold two pieces of information. One is a record you trust. The other is a
claim someone made. You walk the claim one line at a time and ask whether that
line matches the record. Every place the two disagree becomes one plain
sentence.

Python does this with three small parts. `in` asks whether a key is there at
all. `!=` asks whether two values differ. A list collects the sentences you
find.

## The catalogue, and one account

The catalogue is the record the library trusts. It holds one loan length for
each book, counted in days.

The account is what one reader has out right now. Each line on it holds a
title and the number of days that title was lent for.

Two things can be wrong with a line on the account.

1. The book is not in the catalogue at all.
2. The book is in the catalogue, and the loan length does not match.

The code below asks both questions, in that order.

```python
# Sidebar 6: check one account against the catalogue, line by line.
catalogue = {"Salt Marsh Atlas": 21, "Bread Science": 14, "Night Trains": 21}
checked_out = [("Salt Marsh Atlas", 21), ("Bread Science", 28), ("Cloud Charts", 14)]

def loan_gaps(borrowed, record):
    gaps = []
    for book, days in borrowed:
        if book not in record:
            gaps.append(f"{book} is not in the catalogue")
        elif days != record[book]:
            gaps.append(f"{book} was lent for {days} days, catalogue says {record[book]}")
    return gaps

for finding in loan_gaps(checked_out, catalogue):
    print(finding)
```

### What each line does

- Line 1 is a comment. Python ignores it. It is there for you.
- Line 2 makes the catalogue. Three titles, three loan lengths, read by key.
- Line 3 makes the account. It is a list of three pairs. Each pair holds a
  title and a number.
- Line 5 names the function `loan_gaps`. It takes the account as `borrowed`
  and the catalogue as `record`.
- Line 6 makes an empty list. Every problem found goes into it.
- Line 7 starts the loop. Each pass reads one pair from the account. The
  account itself does not change.
- Line 8 asks the first question. Is this title a key in the catalogue?
- Line 9 runs when it is not. It writes one sentence and adds it to `gaps`.
- Line 10 asks the second question. `!=` means is not equal to. It compares
  the days on the account against the days in the catalogue.
- Line 11 runs when those two numbers differ. Its sentence holds both.
- Line 12 hands the whole list back.
- Line 14 calls the function and loops over what came back.
- Line 15 prints one finding per pass.

**How line 7 reads.** Each item in `borrowed` is a pair. Two names in front of
the word `in` open that pair into its two parts. `book` gets the title. `days`
gets the number.

It prints two lines. Bread Science was out longer than the catalogue allows.
Cloud Charts is not in the catalogue at all.

Salt Marsh Atlas says nothing, because its loan length matches. Night Trains
says nothing either, and for a different reason. The loop walks the account,
not the catalogue, so a title nobody borrowed is never looked at.

### Two things to notice

The order of the two questions matters. Line 8 has to come first.

Line 10 reads `record[book]`. Square brackets stop your code with a `KeyError`
when the key is not there. Cloud Charts is not in the catalogue, so line 10
would fail on it. Line 8 catches that case and sends it down its own path, so
line 10 never sees it.

`elif` is what keeps the two paths apart. Exactly one of them runs on each
pass. A title that is missing gets reported once, not twice.

The function hands back a list either way. An empty list is an answer, not a
failure. It says the account and the catalogue agree.
