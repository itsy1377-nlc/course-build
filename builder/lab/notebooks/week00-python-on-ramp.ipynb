## Section 2: Calling functions and using return values

**25 minutes.**

A function is a piece of code with a name. Running it is called calling it.
You call it by writing its name and a pair of brackets.

Most functions hand a value back when they finish. That value is called the
return value. If you do not store it, it is gone.

This is the section week 1 leans on hardest. The example below scales a recipe
for a number of guests.

````python
def scale_recipe(cups, guests):
    return cups * guests / 4

flour = scale_recipe(2, 8)
sugar = scale_recipe(1, 8)

print(flour)
print(sugar)

scale_recipe(2, 12)
print("That last call ran, and nothing kept the answer.")
````

### What each line does

- Lines 1 and 2 make the function. Section 6 covers how to write one. For now
  just call it.
- Line 4 calls it for 2 cups and 8 guests, and stores the answer in `flour`.
- Line 5 calls it again for 1 cup and 8 guests, and stores that in `sugar`.
- Lines 7 and 8 print both stored answers.
- Line 10 calls it a third time. Nothing catches the answer, so the answer is
  lost.
- Line 11 says so on screen.

The one rule to take away: a call gives you a value back, and you have to put
it somewhere.

`flour = scale_recipe(2, 8)` runs the function first, then stores what came
back. Write the call on its own line with no name in front, and the answer
falls on the floor.

````python
# Try changing this.
# Change the number of guests, then run the cell.

butter = scale_recipe(1, 16)
print(butter)
````

## Section 3: Lists, dictionaries and keys

**40 minutes. This section comes in three parts.**

A list holds items in order. You reach an item by its position. Positions
start at 0, not 1.

A dictionary holds values by name. Each name is called a key. You reach a
value by its key.

Ask a dictionary for a key it does not have and your code stops with an error.
So there is also a safe way to ask, with a backup value.

Part one reads. Part two writes. Part three opens a dictionary that holds a
list inside it.

The example below is a library account.


````python
checked_out = ["Dune", "Kindred", "Piranesi"]

print(checked_out[0])
print(checked_out[2])
print(len(checked_out))

checked_out.append("Beloved")
print(checked_out)

account = {"holder": "Rosa Lim", "branch": "Oak Hill", "due_days": 14}

print(account["holder"])
print(account["due_days"])

print(account.get("phone"))
print(account.get("phone", "not on file"))
````

### What each line does

- Line 1 makes a list of three book titles.
- Line 3 prints the first one. Its position is 0.
- Line 4 prints the third one. Its position is 2.
- Line 5 prints how many books are in the list.
- Line 7 adds a fourth book to the end of the list.
- Line 8 prints the whole list, now with four books.
- Line 10 makes a dictionary with three keys.
- Lines 12 and 13 read two values by key.
- Line 15 asks for a key that is not there. `.get()` hands back `None`.
- Line 16 asks the same way, with a backup value to use instead.

`None` is Python's word for no value at all. It is not the text `"None"` and
it is not zero. To ask whether a name holds `None`, you write `x is None`.
Section 7 covers that.

Two ways to read a dictionary, and when to use each.

Use square brackets when the key must be there. Your code stops with a
`KeyError` if it is not, which is honest.

Use `.get()` with a backup when the key may be missing. Your code keeps going
and you get the backup instead.

### Part two: writing to a dictionary

Reading is half the job. You also have to put values in.

Square brackets on the left of an equals sign write a value. The same brackets
on the right read one.

`.update()` folds a second dictionary into the first one.


````python
account["fine_owed"] = 0
print(account)

extra = {"branch": "Live Oak", "card_expires": "June"}

account.update(extra)
print(account)
````

### What each line does

- Line 1 writes a new key. `fine_owed` was not in the account, so this makes it.
- Line 2 prints the account. It now has four keys.
- Line 4 makes a second dictionary with two keys.
- Line 6 folds that second dictionary into the first one.
- Line 7 prints the result.

Two things happen on line 6. `branch` was already a key, so its old value is
replaced. `card_expires` was not there, so it is added.


### Part three: a dictionary that holds a list

A value in a dictionary can be anything. It can be a number, some text, a
list, or another dictionary.

Real data usually stacks that way. A dictionary holds a list, and each item in
that list is its own dictionary.

You reach inside it one step at a time, left to right.


````python
holds = {
    "branch": "Oak Hill",
    "books": [
        {"title": "Dune", "due_days": 14},
        {"title": "Kindred", "due_days": 7},
    ],
}

print(holds["branch"])
print(holds["books"])
print(holds["books"][0])
print(holds["books"][0]["title"])
print(holds["books"][1]["due_days"])
````

### What each line does

- Lines 1 to 7 make one dictionary with two keys.
- The value under `branch` is text. The value under `books` is a list.
- Each item in that list is its own dictionary with two keys.
- Line 9 reads the branch by key. Text comes back.
- Line 10 reads the whole list. Two dictionaries come back.
- Line 11 takes the first item in that list. One dictionary comes back.
- Line 12 goes one step further and reads its `title`. `Dune` comes back.
- Line 13 does the same for the second book and its `due_days`.

Read line 12 from left to right. Open `holds`. Take `books`. Take item 0. Take
`title`. Each step hands the next step something smaller.

Each step has to match what it asks. `[0]` works on a list. `["title"]` works
on a dictionary. Ask a list for a key and Python stops with a `TypeError`.


````python
# Try changing this.
# Add a book, then read a key that is not in the account.

checked_out = ["Dune", "Kindred", "Piranesi", "Beloved"]
checked_out.append("Beloved Again")

print(len(checked_out))
print(account.get("fine_owed", "nothing owed"))
````

## Section 4: JSON, a dictionary in text form

**20 minutes.**

JSON is a way to write a dictionary as plain text. Programs use it to send
data to each other.

JSON looks like a dictionary. It is not one. It is text until you convert it.

`json.loads()` turns JSON text into a dictionary. `json.dumps()` turns a
dictionary back into JSON text.

The example below is one line from a bus arrival board.

````python
import json

line = '{"route": "12", "minutes": 4, "stop": "Elm and 5th"}'

print(line)
print(type(line))

board = json.loads(line)

print(type(board))
print(board["minutes"])
print(board["stop"])

back_to_text = json.dumps(board)
print(back_to_text)
print(type(back_to_text))
````

### What each line does

- Line 1 brings in the `json` tool. It ships with Python.
- Line 3 holds one board line as text. Note the single quotes around it.
- Line 5 prints that text.
- Line 6 prints its kind. `str` means string, which is Python's word for text.
- Line 8 converts the text into a dictionary.
- Line 10 prints its kind. `dict` means dictionary.
- Lines 11 and 12 read two values by key, the way section 3 did.
- Line 14 turns the dictionary back into text.
- Lines 15 and 16 print that text and its kind.

`type()` answers one question: what kind of thing is this. It is the fastest
way to tell text from a dictionary when you are stuck.

### What happens when the text is not JSON

JSON has strict rules. Every key needs double quotes. Every pair needs a
comma. Hand `json.loads()` text that breaks a rule and your code stops.

Here is what that looks like. Do not run this. Read it.

```python
json.loads("route: 12, minutes: 4")
```

```text
json.decoder.JSONDecodeError: Expecting value: line 1 column 1 (char 0)
```

That error is a kind of `ValueError`. The text was a string, which was the
right kind. The content inside it was wrong.

You will see this exact error in week 2. Section 9 shows you how to catch it
and carry on.

````python
# Try changing this.
# Change a value in the text, then convert it and read it back.

line = '{"route": "40", "minutes": 11, "stop": "Main and Pine"}'
board = json.loads(line)
print(board["route"], board["minutes"])
````

## Section 6: Writing your own function

**30 minutes.**

Section 2 called a function somebody else wrote. Now you write one.

Three words do the work. `def` names it. The names in the brackets are its
parameters, and they stand for the values it takes. `return` hands a value
back to whoever called it.

A parameter can carry a default. That is a value to use when the caller does
not give one.

The example below works out how much paint a wall needs.

````python
def paint_needed(width, height, coats=2):
    area = width * height
    return area * coats / 350

gallons = paint_needed(12, 9)
print(gallons)

three_coats = paint_needed(12, 9, 3)
print(three_coats)
````

### What each line does

- Line 1 names the function and lists three parameters. The third has a
  default of 2.
- Line 2 works out the area of the wall and holds it in `area`.
- Line 3 hands back the number of gallons. One gallon covers 350 square feet.
- Line 5 calls it with two values, so `coats` uses its default of 2.
- Line 6 prints that answer.
- Line 8 calls it with three values, so `coats` is 3 instead.
- Line 9 prints that answer.

The lines inside the function are indented by four spaces. That indent is how
Python knows which lines belong to the function. Get it wrong and you get an
`IndentationError`.

The parameter names are only used inside the function. `width` does not exist
outside it.

### The difference between `return` and `print`

This is the most common mix-up in the course. It breaks week 1 on the first
cell.

`print` puts text on the screen. A person reads it.

`return` hands a value back to the code that called the function.

A function with no `return` hands back `None`. Run the cell below. Watch what
happens.

````python
def add_tip(total):
    print(total * 1.2)

answer = add_tip(20)

print("The screen shows 24.0 above. Now look at what came back:")
print(answer)
````

The number 24.0 appeared on screen, so the function clearly worked. But
`answer` holds `None`, because `add_tip` never returned anything.

Later code that tries to use `answer` will fail. Section 10 shows you that
exact error.

Here is the fix. Return the value, and let the caller decide whether to print
it.

````python
def add_tip_fixed(total):
    return total * 1.2

answer = add_tip_fixed(20)
print(answer)
````

## Section 7: if, elif and else

**25 minutes.**

An `if` picks a path. The code under it runs only when a test is true.

`elif` means: if that first test was false, try this one. `else` is the path
taken when every test above it was false.

Exactly one path runs. Python checks them from the top and stops at the first
true one.

The example below is a thermostat.

````python
temp = 41

print(temp > 80)
print(temp > 60)

if temp > 80:
    setting = "cool"
elif temp > 60:
    setting = "off"
else:
    setting = "heat"

print(setting)
````

### What each line does

- Line 1 holds the temperature in a variable.
- Line 3 prints the answer to one test. A test gives back `True` or `False`.
- Line 4 prints the answer to another test.
- Line 6 asks the first question. 41 is not above 80, so its path is skipped.
- Line 8 asks the second. 41 is not above 60, so that path is skipped too.
- Line 10 is the fallback. It runs, so `setting` becomes `heat`.
- Line 13 prints the answer.

The colon at the end of an `if` line is required. So is the indent on the line
below it. Miss either one and Python stops with a `SyntaxError`.

Order matters. Python takes the first true path and ignores the rest. Put the
widest test first and the narrow ones never run.

### Testing whether something is in a list

`in` asks one question: is this value somewhere in that group. It gives back
`True` or `False`.

`not in` asks the opposite.

````python
rooms = ["kitchen", "porch", "garage"]

print("porch" in rooms)
print("attic" in rooms)

if "porch" in rooms:
    print("The porch has a sensor.")

if "attic" not in rooms:
    print("The attic has no sensor.")
````

### Asking whether a name holds nothing

`is` compares one value against another. In this course you use it with
`None`, and almost nowhere else.

`x is None` asks one question. Does `x` hold nothing yet? The answer comes
back as `True` or `False`.

Use `is` for `None`. Use `==` for everything else, such as a number or a word.


````python
reading = None

print(reading is None)

reading = 68

print(reading is None)
````

## Section 8: Loops, in two parts

**35 minutes. This is the longest section.**

Part one is the `for` loop. It does the same thing to each item in a list.

Part two is the `while` loop. It keeps going until a stop condition is met,
with a counter that ends it either way.

Part two is the one week 5 depends on. Give it the extra time.

### Part one: the `for` loop

A `for` loop takes each item in a list. It takes one at a time. It runs the
same lines on each one.

The example below is a class roster of quiz scores. It builds one line of text
per student. Then it joins those lines into one block.

````python
roster = [
    {"name": "Ana", "score": 88},
    {"name": "Ben", "score": 72},
    {"name": "Cruz", "score": 95},
]

lines = []

for student in roster:
    lines.append(f"{student['name']}: {student['score']}")

report = "\n".join(lines)
print(report)
````

### What each line does

- Lines 1 to 5 make a list of three dictionaries. Each one has two keys.
- Line 7 makes an empty list to collect the finished lines.
- Line 9 starts the loop. `student` holds one dictionary at a time.
- Line 10 builds one line of text and adds it to `lines`.
- Line 12 joins the three lines into one block, with a line break between them.
- Line 13 prints the block.

Three things to notice.

`student` is a name you chose. Python fills it for you, once per pass.

The lines under `for` are indented. That indent is what puts them inside the
loop.

`"\n"` means a line break. `.join()` puts it between each item, so three
lines come out stacked.

````python
# A for loop can also add numbers up as it goes.

total = 0

for student in roster:
    total = total + student["score"]

print(total)
print(total / len(roster))
````

````python
# Try changing this.
# Add a fourth student, then run the cell just above this one again.
# The total changes from 255 to 316. The average changes from 85.0 to 79.0.

roster = [
    {"name": "Ana", "score": 88},
    {"name": "Ben", "score": 72},
    {"name": "Cruz", "score": 95},
    {"name": "Dana", "score": 61},
]

for student in roster:
    print(student["name"], student["score"])
````

### Part two: the `while` loop, and the counter that ends it

A `for` loop always ends, because a list runs out.

A `while` loop is different. It keeps going while its test stays true. If the
test never turns false, the loop never ends.

Here is a loop with no way out. **Do not run this.** Read it.

```python
found = None
while found is None:
    print("still looking")
```

Nothing inside that loop can ever change `found`, so it prints forever.

The fix is a counter. You count the passes, and you stop at a cap no matter
what else happens.

The example below reads quiz scores until it finds one below passing. It stops
anyway after three passes.

````python
passing = 75
scores = [88, 72, 95, 61, 79]

steps = 0
max_steps = 3
found = None

while found is None and steps < max_steps:
    score = scores[steps]
    print("pass", steps + 1, "reads", score)
    if score < passing:
        found = score
    steps = steps + 1

print("passes taken:", steps)
print("first low score:", found)
````

### What each line does

- Line 1 sets the passing mark.
- Line 2 holds five scores.
- Line 4 starts the counter at 0.
- Line 5 sets the cap. The loop can run three times at most.
- Line 6 sets `found` to `None`, which means nothing found yet.
- Line 8 is the test. Both parts must be true for the loop to run again.
- Line 9 reads the score at the counter's position.
- Line 10 prints what this pass saw.
- Lines 11 and 12 store the score if it is below passing.
- Line 13 adds one to the counter. This line is what ends the loop.
- Lines 15 and 16 print the result.

The test on line 8 has two parts, and either one can stop the loop.

`found is None` is the stop condition. It turns false the moment the loop finds
what it came for.

`steps < max_steps` is the cap. It turns false after three passes, whether or
not anything was found.

Forget line 13 and the counter never moves. That is the runaway loop again.

````python
# Watch the cap do its job.
# No score in this list is below passing, so only the cap can end the loop.

passing = 75
scores = [88, 92, 95, 81, 79]

steps = 0
max_steps = 3
found = None

while found is None and steps < max_steps:
    print("pass", steps + 1, "reads", scores[steps])
    if scores[steps] < passing:
        found = scores[steps]
    steps = steps + 1

print("passes taken:", steps)
print("first low score:", found)
````

## Section 9: try and except

**25 minutes.**

Some lines can fail for reasons you cannot prevent. Text arrives that is not a
number. A key is missing. A conversion does not work.

`try` runs the risky lines. `except` names one error type and says what to do
when it happens. Your program keeps going.

The example below splits a bill.

````python
text = "eight dollars"

try:
    amount = float(text)
    print("converted:", amount)
except ValueError:
    print("That is not a number. Using 0 instead.")
    amount = 0.0

print("amount is", amount)
````

### What each line does

- Line 1 holds an entry that looks like money but is not a number.
- Line 3 starts the risky part.
- Line 4 tries to turn the text into a number. This line fails.
- Line 5 never runs, because line 4 stopped.
- Line 6 catches the failure. `ValueError` is the error type line 4 raises.
- Lines 7 and 8 run instead. They report the problem and set a backup value.
- Line 10 runs either way, because the program never stopped.

Name the error type you expect. Writing `except Exception:` catches every
error, including the ones you did not plan for.

That hides the message section 10 teaches you to read, and it turns a clear
failure into a silent wrong answer.

````python
# One bad entry should not stop the whole bill.
# Put the try inside a loop and the loop carries on.

entries = ["12.50", "8.75", "eight dollars", "20.00"]

total = 0.0

for entry in entries:
    try:
        total = total + float(entry)
    except ValueError:
        print("skipped:", entry)

print("total:", round(total, 2))
````

### Trying again, up to a limit

Sometimes the answer is not to skip the failure. It is to ask again.

Put the `try` inside a `while` loop with a counter, the way section 8 did.
That gives you a limited number of tries.

The loop needs an accepted value, not merely a value that converted. It loads
each try into `candidate`. It assigns `value` only after the candidate
passes the check. The variable in the `while` condition therefore stays
`None` until one try is accepted.

This is the exact shape of the week 2 retry: load a candidate, check it, and
stop only when the check passes.

````python
readings = ["not a number", "900", "41.2"]

tries = 0
value = None

while value is None and tries < 3:
    candidate = None
    try:
        candidate = float(readings[tries])
    except ValueError:
        print("try", tries + 1, "did not hold a number")
    if candidate is not None:
        if candidate >= -100 and candidate <= 150:
            value = candidate
        else:
            print("try", tries + 1, "was outside the allowed range")
    tries = tries + 1

print("tries used:", tries)
print("accepted value:", value)
````

````python
# Try changing this.
# Every entry below is unreadable or outside the allowed range.
# Watch the cap stop the loop with no accepted value.

readings = ["a", "900", "-500"]

tries = 0
value = None

while value is None and tries < 3:
    candidate = None
    try:
        candidate = float(readings[tries])
    except ValueError:
        print("try", tries + 1, "did not hold a number")
    if candidate is not None:
        if candidate >= -100 and candidate <= 150:
            value = candidate
        else:
            print("try", tries + 1, "was outside the allowed range")
    tries = tries + 1

print("tries used:", tries)
print("accepted value:", value)
````

## Section 10: How to read an error message

**30 minutes. This is the most useful section in the notebook.**

It is 11 at night. Your lab is due Sunday. A cell just failed, and you are the
only person in the room.

This section is the answer to that. Read it now. Come back to it every week.

Python hides nothing. It prints what went wrong and where. It just prints the
answer at the bottom, and most people read from the top.

The section comes in two parts. Part one is the routine, and how a traceback
is laid out. Part two walks six errors, one at a time.


### Part one: the routine, five steps

1. **Read the last line first.** Everything above it is context.
2. **Name the error type.** It is the word before the colon on that last line.
3. **Find your line.** Look for the last line that names your own cell.
4. **Print the value** on the line above the one that failed.
5. **Change one thing.** Run it again.

Step 1 is the one that changes outcomes. Students read from the top, hit file
paths from code they did not write, and decide the message is unreadable.

It is not. Start at the bottom.

### The anatomy of a traceback

Here is a traceback from a small shopping program. Four parts are labeled
below it.

```text
Traceback (most recent call last):
  File "shopping.py", line 12, in <module>
    print(basket_total(basket))
  File "shopping.py", line 7, in basket_total
    return sum(basket["prices"])
KeyError: 'prices'
```

**1. The header line.** `Traceback (most recent call last)`. It says a block of
error text follows, and it tells you the reading order.

**2. The frames.** Each `File` line and the code line under it is one frame. A
frame is one step in the chain of calls. This traceback has two.

**3. The last frame.** `line 7, in basket_total`, and the line of code under
it. That is the line that actually failed. It is the last frame, just above
the error.

**4. The error type and its message.** `KeyError: 'prices'`. The type is the
word before the colon. The message is what comes after it.

Some Python versions also draw a small mark under the exact part of the line
that failed. Your notebook server may or may not show one.

### Why the useful part is at the bottom

The header says "most recent call last". That is Python telling you the
reading order.

Your program called one function. That function called another. Each call is a
frame, and Python prints them oldest first.

So the last frame was running when everything stopped. That is where the
problem is. The error sits right below it.

Read the bottom two lines and you have the type, the message and the failing
line. Read up only to see how the code got there.

### Part two: six errors, and how to fix each one

Five of the six errors below get two cells. The first breaks on purpose. The
second fixes it. Read the cell, predict the error, then run it.

The sixth error cannot be run at all. Its own part says why.

**How these cells are written.** Each broken cell is wrapped, so the notebook
runs to the end. The message you see is the real one.

The wrappers catch every error on purpose, because each cell raises a
different one. Section 9 still holds. In your own code, name the type you
expect.

One more thing. In a notebook the `File` line names a long temporary file. That
file is the cell you just ran.

Run the cell below first. It sets up the shopping basket.


````python
import traceback


def show_the_error():
    """Print the error Python just raised, then let the notebook carry on."""
    print(traceback.format_exc())


basket = {"apples": 3, "bread": 1}
print(basket)
````

#### `NameError`

A name is used before it exists. Most often it is a typo, or a cell you never
ran.

Predict the error, then run the cell.

````python
try:
    print(bascket)
except Exception:
    show_the_error()
````

The last line names the error: `NameError: name 'bascket' is not defined`.
Newer versions of Python add a guess after it, such as `Did you mean:
'basket'?`.

Python is not saying the basket is empty. It is saying it has never heard of
`bascket` at all. Look at the spelling.

The fix is below.

````python
print(basket)
````

#### `KeyError`

A dictionary was asked for a key it does not have.

Predict the error, then run the cell.

````python
try:
    print(basket["pears"])
except Exception:
    show_the_error()
````

The last line reads `KeyError: 'pears'`.

The quoted word is the key you asked for. It is not in the dictionary.

Two fixes. Print the dictionary and check what keys it really has. Or use
`.get()` with a backup, the way section 3 showed.

````python
print(basket.keys())
print(basket.get("pears", 0))
````

#### `TypeError`

A value was used in a way its kind does not allow.

The most common case in this course is a `None` where a dictionary was
expected. Section 6 showed how a function ends up handing back `None`.

Predict the error, then run the cell.

````python
def find_basket(store_name):
    if store_name == "corner store":
        return {"apples": 3}
    # Every other name falls off the end, so this hands back None.


result = find_basket("night market")

try:
    print(result["apples"])
except Exception:
    show_the_error()
````

The last line reads `TypeError: 'NoneType' object is not subscriptable`.

In plain words: you used square brackets on `None`, and `None` does not take
square brackets.

The failing line is not the wrong line. The wrong line is the one that set
`result`. Print `result` and you see it at once.

Check for `None` before you use the value.

````python
result = find_basket("night market")
print("result is:", result)

if result is None:
    print("No basket at that store. Nothing to read.")
else:
    print(result["apples"])
````

#### `ValueError`, including the JSON one

The kind was right and the content was wrong. Text that is not valid JSON is
the case you will meet.

Predict the error, then run the cell.

````python
import json

text = "apples: 3, bread: 1"

try:
    loaded = json.loads(text)
except Exception:
    show_the_error()
````

The last line reads `json.decoder.JSONDecodeError: Expecting value: line 1
column 1 (char 0)`.

That error is a kind of `ValueError`. The message points at the first
character that broke the rules.

This one is worth a second look. Its traceback has four frames, and three of
them are inside Python's own JSON code. Only the first frame is yours.

That is normal, and it is why step 3 says find your line. Read the bottom
line, then look for the frame that names your cell.

JSON needs double quotes around every key and around every piece of text. The
fix is below.

````python
text = '{"apples": 3, "bread": 1}'
loaded = json.loads(text)
print(loaded)
print(loaded["apples"])
````

#### `AttributeError`

You asked a value for something it does not have. Often it is a text move used
on a list.

Predict the error, then run the cell.

````python
items = ["apples", "bread"]

try:
    print(items.upper())
except Exception:
    show_the_error()
````

The last line reads `AttributeError: 'list' object has no attribute 'upper'`.

The message names the kind you have, `list`, and the thing you asked for,
`upper`.

`.upper()` works on text. It does not work on a list. Work on one item, or
join the list into one string first.

````python
print(items[0].upper())
print(", ".join(items).upper())
````

#### `SyntaxError` and `IndentationError`

These two are different from the other four. Python could not read the line at
all, so nothing ran.

That is why there is no runnable cell here. `try` and `except` cannot catch a
line Python never got to run.

Here is what a `SyntaxError` looks like. The colon is missing.

```python
if hours > 4
    print("long stay")
```

```text
  File "meter.py", line 1
    if hours > 4
                ^
SyntaxError: expected ':'
```

And here is an `IndentationError`. The colon is there, and the line below it
is not indented.

```python
if hours > 4:
print("long stay")
```

```text
  File "meter.py", line 2
    print("long stay")
    ^
IndentationError: expected an indented block after 'if' statement on line 1
```

Both point at a line number and draw a mark under the spot. Check three things
in this order: a missing colon, an unclosed bracket or quote, and the indent.

An unclosed bracket often reports the line **after** the real problem, because
Python kept reading, looking for the closing bracket.

### The notebook trap

This one is specific to notebooks. It costs students more time than anything
else here.

The cell that shows the error is not always the cell that is wrong.

Cells run in the order you click them. They do not run in the order they sit
on the page.

A name made in a cell you never ran does not exist. You can see it on the
screen. Python has still never heard of it.

So a `NameError` can point at a cell that is perfectly correct.

The fix, in order.

1. Run the cells above the one that failed, from the top.
2. If that does not do it, restart the notebook and run every cell in order.
3. Then read the error again.

Restarting clears every name and starts clean. Do it when you come back to a
lab after a break.

### Three things not to do

1. **Do not retype the whole cell.** The message already names the line that
   failed. Fix that line.
2. **Do not delete the error and try something else at random.** Read it
   first. Random changes turn one clear error into three unclear ones.
3. **Do not paste your whole notebook into an AI assistant before you read the
   last line.** The course AI policy allows AI help with disclosure. It works
   far better when you can say which error you got and which line raised it.

One more habit, and it is the useful one. Print the value on the line above
the failure. Most errors in this course come from a value not being what you
thought it was.
