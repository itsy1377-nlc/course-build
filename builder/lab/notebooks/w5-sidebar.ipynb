## The route

A bus runs a route of four stops. At each stop you write down how many minutes
late the bus is. One leg is the ride from one stop to the next.

A cap is a limit on how many passes a loop takes. This ride is capped at three
legs. A stop condition ends a loop early. This one fires when the bus is back
on time.

Run the cell below.

```python
# Sidebar 5: ride one leg at a time, and stop when a rule says stop.
timetable = {"Oak St": 4, "Vine St": 0, "Rail Yard": 2, "Depot": 3}
route = ["Oak St", "Vine St", "Rail Yard", "Depot"]
called_out = []
legs = 0
MAX_LEGS = 3
finish_note = "rode every leg the cap allowed"
while legs < MAX_LEGS and legs < len(route):
    stop_name = route[legs]
    minutes_late = timetable[stop_name]
    called_out.append(stop_name)
    legs = legs + 1
    print(f"leg {legs}: {stop_name}, {minutes_late} minutes late")
    if minutes_late == 0:
        finish_note = f"stopped at {stop_name}, the bus was back on time"
        break
print(f"legs ridden: {legs}")
print(f"stops called out: {called_out}")
print(f"why it ended: {finish_note}")
```

### What each line does, before the loop

- Line 1 is a comment. Python ignores it. It is there for you.
- Line 2 holds one number for each stop: how many minutes late the bus is
  there.
- Line 3 is the route, in order.
- Line 4 starts an empty list. It fills up as the ride goes on.
- Line 5 starts the counter at zero.
- Line 6 sets the cap. `MAX_LEGS` never changes.
- Line 7 holds the reason the ride ended, so there is always a note to print.

### What each line does, inside the loop and after

- Line 8 is the loop test. `while` means keep going while both tests are true.
  `legs < MAX_LEGS` is the cap. `legs < len(route)` is true while a stop is
  still ahead, because `len(route)` is how many stops the route holds. When
  either test fails, the loop ends.
- Line 9 reads the stop at position `legs`. The counter is the position, so
  the first pass reads position 0, which is Oak St. This is the same line
  week 0 wrote as `score = scores[steps]`.
- Lines 10 and 11 read the minutes for that stop and add the stop to the list
  already called out.
- Line 12 moves the counter. Without this line the cap never fires, and line 9
  reads the same stop forever.
- Line 13 prints one line per pass.
- Lines 14 and 15 are the stop condition. The bus is on time when
  `minutes_late` is zero, so the note is rewritten.
- Line 16 ends the loop at once. `break` is the word that does it.
- Lines 17, 18 and 19 run once, after the loop.

Two lines end this ride, and they do different jobs. Line 8 ends it on the
cap. Line 16 ends it on the stop condition. The lab needs both.

## The counter that never moves

Line 12 is the line that moves the counter. Take it out and `legs` stays at
zero. The first test never turns false. Line 9 reads the same stop every pass,
so the second test never turns false either.

A loop like that runs until you stop the notebook by hand. Week 0 shows one
and tells you not to run it.

The cell below leaves the counter line out on purpose, and it is safe to run.
It carries a second counter that ends it after eight passes. That second
counter is a hard stop. A hard stop is not a cap. It is there so the cell
finishes.

Run it, then read how many legs the counter recorded.

```python
# The counter line is missing on purpose. Safe to run: a hard stop ends it.
stalled_legs = 0
stalled_cap = 3
passes_taken = 0
HARD_STOP = 8
while stalled_legs < stalled_cap and passes_taken < HARD_STOP:
    passes_taken = passes_taken + 1
    print(f"pass {passes_taken}: {route[stalled_legs]}, counter still {stalled_legs}")
print(f"passes taken: {passes_taken}")
print(f"legs the counter recorded: {stalled_legs}")
print("The counter never moved, so neither test could turn false.")
print("The hard stop ended this one. A counter you never move is not a cap.")
```

## `break` and `return` end different things

`break` ends the loop. The lines under the loop still run.

`return` inside a function ends the function. The lines under the loop do not
run at all.

The lab needs `return`. Here is why. The lab's loop sets the reason the run
ended. The line under that loop then sets the reason again, for a run that
nothing settled. A `break` leaves the loop and lets that line write over the
reason you just set. A `return` leaves the function before that line can run.

The cell below is the same ride, moved inside a function. It has two ways out.
One is the `return` inside the loop. One is the `return` under it. Read both
before you run it.

```python
# The same ride, inside a function. Two ways out, and both are a return.
def ride_route(cap):
    """Ride the route and hand back one note saying why the ride ended."""
    legs_ridden = 0
    while legs_ridden < cap and legs_ridden < len(route):
        stop_name = route[legs_ridden]
        minutes_late = timetable[stop_name]
        legs_ridden = legs_ridden + 1
        print(f"leg {legs_ridden}: {stop_name}, {minutes_late} minutes late")
        if minutes_late == 0:
            return f"stopped at {stop_name}, the bus was back on time"
    return "rode every leg the cap allowed"


print("cap of 3, why it ended:", ride_route(3))
print()
print("cap of 1, why it ended:", ride_route(1))
```

### What the two runs showed

The first run reached Vine St, where the bus is on time. The `return` inside
the loop handed the note back from there, and the last line of the function
never ran.

The second run had a cap of one. The stop condition never fired, so the loop
test ended it and the last line of the function handed back the other note.

Change the inner `return` to `break` and run it again. Both runs then give
back the same note, because the last line runs either way. That is the mistake
the lab's check cell watches for.
