## 2A. A reading is a dictionary

One reading holds three facts. Which station sent it. How warm it was. What
the sky looked like.

A dictionary holds values under names. Each name is called a key. You read a
value out by naming its key in square brackets.

That works only when the key is there. When a key might be missing, use
`get()` instead.

```python
# 2A. One rooftop reading, held in a dictionary.
reading = {"station": "Rooftop A", "temp_f": 71.4, "sky": "clear"}

print("The whole reading:", reading)
print("Station:", reading["station"])
print("Temperature:", reading["temp_f"])

# The station sent no wind speed this hour. get() answers without stopping.
print("Wind:", reading.get("wind_mph"))
print("Wind, with a stand-in:", reading.get("wind_mph", "not recorded"))

# You can add a key after you make the dictionary.
reading["wind_mph"] = 6.2
print("Wind now:", reading["wind_mph"])
print("Every key in this reading:", list(reading))
```

### What each line does

- Line 2 makes the dictionary. Three keys, three values, inside curly braces.
- Line 4 prints the whole thing. Python shows every key with its value.
- Lines 5 and 6 read one value each, by naming its key.
- Line 9 asks for a key that is not there. `get()` hands back `None`.
- Line 10 asks again with a stand-in value. It gets that instead of `None`.
- Line 13 adds a new key. The dictionary is four keys long now.
- Line 15 lists the key names on their own, without the values.

## 2B. The same reading, arriving as text

The station does not hand you a dictionary. It sends one line of text. The
text looks like a dictionary. Python still treats it as one long string.

That text format is called JSON. `json.loads()` turns a line of JSON into a
real dictionary. Only after that can you read it by key.

```python
# 2B. The line the station sends, and what it takes to use it.
import json

gauge_line = '{"station": "Rooftop A", "temp_f": 71.4, "sky": "clear"}'

print("What arrived:", gauge_line)
print("Python calls it:", type(gauge_line))
print("Its first character:", gauge_line[0])

# Text that looks like a dictionary is not one yet.
reading = json.loads(gauge_line)

print("After json.loads():", reading)
print("Python calls it now:", type(reading))
print("Temperature:", reading["temp_f"])
print("Half a degree warmer:", reading["temp_f"] + 0.5)
```

### What each line does

- Line 2 brings in the `json` tool. It ships with Python.
- Line 4 holds the line the station sent, in `gauge_line`. Note the quotes
  around all of it.
- Line 7 shows what Python thinks it is. The answer is a string.
- Line 8 prints character zero. It is a curly brace, not a key.
- Line 11 turns the text into a dictionary.
- Line 14 shows what it is now. The answer is a dict.
- Lines 15 and 16 read a value and do arithmetic on it. Line 11 is what made
  that possible.

## 2C. When the line is bad, ask again

The gauge on the roof is not perfect. Now and then it sends a line you cannot
read. Sometimes the line stops early. Sometimes it is not JSON at all.

A bad line is normal. It is not a reason to stop. You catch the error, then
ask for the next reading.

`try` and `except` do that. Python runs the lines under `try`. When one of
them fails, it jumps to `except` instead of stopping the whole cell.

A counter ends the loop either way. Three tries is enough.

```python
# 2C. The gauge sends a bad line sometimes. Ask again, up to three times.
import json

LINES = ['{"station": "Rooftop A", "temp_f": 68', 'sensor warming up',
         '{"station": "Rooftop A", "temp_f": 69.8}']

reading = None
attempts_made = 0

while reading is None and attempts_made < 3:
    gauge_line = LINES[attempts_made]
    attempts_made = attempts_made + 1
    print("Try", attempts_made, "got:", gauge_line)
    try:
        reading = json.loads(gauge_line)
    except json.JSONDecodeError as error:
        print("   That line was bad:", error)

print("Tries used:", attempts_made, "| reading after the loop:", reading)
```

### What each line does

- Lines 4 and 5 hold three lines from the gauge. The first stops early. The
  second is a plain sentence. The third is good.
- Line 7 sets `reading` to `None`. That is the answer you are left with if
  every try fails.
- Line 8 starts the counter `attempts_made` at zero.
- Line 10 is the loop test. It keeps going while there is no reading and the
  counter is under three.
- Line 11 takes the next line from `LINES` and holds it in `gauge_line`.
- Line 15 is the line that can fail. It sits under `try`.
- Lines 16 and 17 catch the failure and print it. The loop then goes round
  again.
- Line 19 runs whether the loop found a reading or not.

The loop stops the moment `reading` holds something. Nothing has to break it.
