## 4A. A function that hands a number back

A function is a piece of code with a name. You hand it values. It hands one
answer back.

The values you hand it are its parameters. The answer it hands back is its
return value. `return` is the word that does the handing.

A function that prints is not the same as a function that returns. A printed
number is on the screen and nowhere else. A returned number is a value your
next line can use.

Below, two functions read the same shelf of playlists. One counts songs. The
other turns that count into minutes.

```python
# Sidebar 4A: two functions that hand a number back.
# Each song runs about four minutes, so one count turns into minutes.
MINUTES_PER_SONG = 4

music_shelf = {"road trip": ["Gravel Turnoff", "Six Lanes Open",
                             "Night Windows", "Two Hours Out"],
               "study": ["Quiet Hour", "Second Draft", "Paper Lamp"]}


def count_songs(playlist_title, shelf):
    return len(shelf[playlist_title])


def playing_minutes(playlist_title, shelf):
    return count_songs(playlist_title, shelf) * MINUTES_PER_SONG


print("Road trip songs:  ", count_songs("road trip", music_shelf))
print("Road trip minutes:", playing_minutes("road trip", music_shelf))
print("Study songs:      ", count_songs("study", music_shelf))
print("Study minutes:    ", playing_minutes("study", music_shelf))
```

### What each line does

- Line 3 stores one number, the minutes one song runs. A name in capital
  letters is the usual way to mark a value you do not plan to change.
- Lines 5 to 7 build the dictionary `music_shelf`. Each key is a playlist
  title. Each value is a list of song titles.
- Line 10 names a function, `count_songs`. It takes two values. The first is
  the playlist title. The second is the shelf to look in.
- Line 11 reads the list stored under that title, counts it with `len()`, and
  hands the count back. `return` is what hands it back.
- Line 14 names a second function, `playing_minutes`. It takes the same two
  values.
- Line 15 calls `count_songs`, then multiplies that answer by the number on
  line 3. One function can use what another function hands back.
- Lines 18 to 21 call the two functions on both playlists. Same code, four
  answers.

Two things to notice.

Neither function prints anything. Each one hands a number back, and the print
lines at the bottom decide what to do with it. That is what lets line 15 use
`count_songs` inside a multiplication.

The shelf arrives as a parameter. `count_songs` never names `music_shelf`
inside itself, so it counts any shelf you hand it.

## 4B. Pick the function from a name

Now someone else picks the report. They do not call your two functions. They
hand you a name and wait for an answer.

Your code reads the name and runs the function that matches it. That is the
whole idea. `if` and `elif` do the reading. `else` covers every name that
matches nothing.

That last part matters most. A name you do not know is not a reason to stop.
Hand a sentence back instead, and the code that called you keeps running.

Run part 4A before this cell. The two functions come from there.

```python
# Sidebar 4B: pick which function to run from a name.
# Run the 4A cell first. This cell uses the two functions from it.
def run_report(report_title, playlist_title):
    if report_title == "count":
        return count_songs(playlist_title, music_shelf)
    elif report_title == "minutes":
        return playing_minutes(playlist_title, music_shelf)
    else:
        return "There is no report named " + report_title + "."


print("count on road trip:  ", run_report("count", "road trip"))
print("minutes on road trip:", run_report("minutes", "road trip"))
```

### What each line does

- Line 3 names the function `run_report`. It takes two pieces of text: which
  report to run, and which playlist to run it on.
- Line 4 asks whether the report title is the word `count`. Two equals signs
  ask a question. One equals sign stores a value.
- Line 5 runs `count_songs` and hands that answer straight back. The function
  stops there. Nothing below line 5 runs.
- Line 6 asks the same question about the word `minutes`. Python reaches this
  line only when the answer on line 4 was no.
- Line 7 runs the other function and hands that answer back.
- Line 8 covers every other title.
- Line 9 hands a sentence back. The title matched nothing, and the function
  still answers.
- Lines 12 and 13 ask for both reports on the road trip playlist.

Two things to notice.

`run_report` counts nothing and multiplies nothing. It picks which function
does the work. The two functions from 4A did not change at all.

Whoever calls `run_report` no longer needs the two function names. They need
the report titles, which are ordinary text.

```python
# Try changing this. Nothing here is graded.
#
# The first line below asks for a report that does not exist. Run the cell
# and read what comes back.
#
# Then change "length" to another title that does not exist, and run it
# again. Ask yourself one question. Your code got a sentence back, and the
# second line still ran. What would that second line have done if the
# function had stopped the notebook instead?
print("length report:", run_report("length", "road trip"))
print("count report: ", run_report("count", "road trip"))
```
