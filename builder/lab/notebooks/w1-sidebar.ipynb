## Part 1: call a function and keep what it hands back

**About 15 minutes.**

A stockroom clerk counts stock in cases. Each case holds six boxes. A function
turns cases into boxes.

Run the cell below.


```python
# Sidebar 1A: call a function and keep what it hands back
def boxes_in_cases(cases):
    return cases * 6

morning_boxes = boxes_in_cases(4)
afternoon_boxes = boxes_in_cases(9)
day_total = morning_boxes + afternoon_boxes

print(morning_boxes)
print(afternoon_boxes)
print(day_total)
```

### What each line does

- Line 2 names a function, `boxes_in_cases`. It takes one number, `cases`.
- Line 3 hands a number back. That is what `return` does.
- Line 5 calls the function with 4. The answer is stored in `morning_boxes`.
- Line 6 calls it again with 9. The answer is stored in `afternoon_boxes`.
- Line 7 adds the two stored answers together.
- Lines 9, 10 and 11 print all three numbers.

Two things to notice.

A call hands one value back. If you do not store it, it is gone. Line 5 stores
it, so line 7 can still use it.

The function ran twice with two different numbers. Same code, two answers.
That is what a parameter is for.

**Try changing this.** Change the 9 on line 6 to a 2. Predict all three
printed lines before you run the cell again.

## Part 2: build a sentence from values with an f-string

**About 10 minutes.**

The same clerk writes a shelf note. The note carries three values inside one
sentence. An f-string builds it.

Run the cell below.


```python
# Sidebar 1B: build a sentence from values with an f-string
shelf_name = "rice"
on_hand = 4
low_mark = 5
note = f"{shelf_name}: {on_hand} boxes on hand, reorder at {low_mark}."
print(note)
print(f"Short by {low_mark - on_hand} box.")
```

### What each line does

- Lines 2, 3 and 4 store three values.
- Line 5 builds one sentence and stores it in `note`.
- The `f` in front of the quote is what makes the braces work.
- Anything inside braces is worked out first. Its value then drops into the
  sentence.
- Line 6 prints the finished sentence.
- Line 7 builds a sentence and prints it in one step. The braces hold a
  subtraction, so Python works that out first.

Take the `f` away from line 5 and run it again. The braces print as braces.

## Part 3: a dictionary, a list inside it, and `.update()`

**About 15 minutes.**

The clerk writes an order. An order is not one value. It is several values with
names, and one of those names holds a list.

The lab builds a request the same way. Read this part before you write it.

Run the cell below.


```python
# Sidebar 1C: a dictionary, a list inside it, and .update()
order = {"shelf": "rice", "cases": 4}
order["lines"] = [{"item": "long grain", "boxes": 24}]

print(order["shelf"])
print(order["lines"][0]["item"])

extras = {"delivery": "Tuesday"}
order.update(extras)

print(order)
print(order["lines"][0])
```

### What each line does

- Line 2 builds a dictionary. Each name in it is a **key**, and each key holds a
  value. `order` has two keys so far.
- Line 3 adds a third key, `lines`. Its value is a list, and that list holds one
  dictionary of its own. So there are now two dictionaries: the outer one, and
  the one inside the list.
- Line 5 reads a value off the outer dictionary.
- Line 6 reaches into the inner one. Read it left to right: take `lines`, take
  the first thing in that list, take its `item`.
- Line 8 builds a second dictionary with one key in it.
- Line 9 merges that dictionary into `order`. `.update()` adds every key it is
  handed to the dictionary it is called on.
- Lines 11 and 12 print both dictionaries so you can see where `delivery` went.

**The thing to see here.** `.update()` acts on the dictionary you call it on and
on no other. Line 9 calls it on `order`, so `delivery` lands on the order. The
dictionary inside the list is untouched.

**Try changing this.** Move the merge onto the inner dictionary. Change line 9
to `order["lines"][0].update(extras)`, then run the cell again and read both
printed dictionaries. `delivery` moves off the order and onto the line.

Nothing errors either way. Both are valid Python and both print happily. That is
why this is worth doing once by hand: when a value lands in the wrong
dictionary, nothing tells you.
