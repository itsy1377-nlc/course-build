## One block of text, built one part at a time

You are not printing one value here. You are building a longer piece of text
out of several smaller parts.

Three things decide what that text looks like.

- The values themselves. Those sit in a dictionary.
- The word printed in front of each value. Those sit in a second dictionary,
  under the same keys.
- The order the lines come out in. That is a list of keys, kept apart from both
  dictionaries.

**Keeping the order in its own list is what makes the order yours to set.** You
change the list and the block changes. You touch no value and no heading.

Run the cell below.

```python
# Build one roster line out of several fields, in an order you choose.
roster = {
    "surname": "Okonkwo",
    "given": "Amara",
    "seat": 4,
}
headings = {"surname": "Last", "given": "First", "seat": "Seat"}
line_order = ["given", "surname", "seat"]

roster_lines = []
for field in line_order:
    roster_lines.append(headings[field] + ": " + str(roster[field]))

roster_block = "\n".join(roster_lines)
print(roster_block)
```

### What each line does

- Lines 2 to 6 make a dictionary named `roster`. It holds three details about
  one student, each under a key. Two of them are text. The seat is a number.
- Line 7 makes a second dictionary named `headings`. It holds the word printed
  in front of each detail. It uses the same three keys.
- Line 8 makes a list named `line_order`. It holds those keys in the order the
  lines come out. `given` is first here, and `surname` is second.
- Line 10 makes an empty list. The finished lines go into it, one at a time.
- Line 11 starts the loop. `field` holds one key at a time, in the order line 8
  set.
- Line 12 builds one line and adds it to `roster_lines`. It joins three pieces
  with `+`. The heading for that key, then a colon and a space, then the value
  for that key.
- Line 12 also wraps the value in `str()`. A `+` joins text to text, and the
  seat is a number. `str()` turns any value into text. Take the `str()` off
  and Python stops with a `TypeError`.
- Line 14 turns the three collected lines into one piece of text. `"\n"` means
  a line break, and `.join()` puts one between each item.
- Line 15 prints that text.

Two things to notice.

The lines print in the order `line_order` sets, not the order the keys were
written in `roster`. Read the printed block against line 8 and you can see it.

`headings[field]` and `roster[field]` read the same key out of two different
dictionaries. That is why the second dictionary has to use the same keys as the
first.

```python
# The same work, in two halves. Look at the list first, then at the block.
print("Before the join, roster_lines is a list:")
print(roster_lines)
print("Python calls it:", type(roster_lines))
print("It holds", len(roster_lines), "items.")

print()
print("After the join, roster_block is one piece of text:")
print(roster_block)
print("Python calls it:", type(roster_block))
print("It holds", len(roster_block), "characters.")
```

### What that cell showed

The loop did not build the block. It built a list of three separate lines.

`.join()` is the step that turns those three lines into one piece of text.
Before it you have a list, and you can count the items in it. After it you have
text, and you can print it or count the characters in it.

The lab counts the characters in the block you build there. That count is the
size of what you are sending, so it is worth a look.
