## 7A. Count the bakes that matched the plan

Four bakes sit in one list. Each bake is a dictionary with two keys.
`planned` is the number of rolls the recipe said. `made` is the number you
took out of the oven.

A `for` loop takes one bake at a time. Inside the loop you ask one question
about that bake. When the answer is yes, you add one to a counter.

A counter is a plain number that starts at zero. Nothing about it is
special. The loop is what makes it useful.

```python
# 7A. A yield log holds four bakes. Each bake has the rolls the recipe
# planned, and the rolls you got. Count the bakes that matched the plan.
BAKES = [{"planned": 24, "made": 24},
         {"planned": 24, "made": 21},
         {"planned": 36, "made": 36},
         {"planned": 12, "made": 15}]

def count_on_target(bakes):
    on_target = 0
    for bake in bakes:
        print("planned", bake["planned"], "and made", bake["made"])
        if bake["made"] == bake["planned"]:
            on_target = on_target + 1
    return on_target

print("Bakes that matched the planned number:", count_on_target(BAKES))
```

### What each line does

- Lines 3 to 6 make the list. Four dictionaries, two keys each.
- Line 8 names the function. It takes the list, under the name `bakes`.
- Line 9 starts the counter at zero, before the loop runs.
- Line 10 starts the loop. `bake` holds one dictionary at a time.
- Line 11 prints the two numbers for that bake, so you can watch the loop
  work.
- Line 12 asks the one question. Two equals signs test whether two values
  match.
- Line 13 adds one to the counter. It runs only when line 12 is true.
- Line 14 hands the finished count back.
- Line 16 calls the function and prints what came back.

Two things to notice.

Line 9 sits outside the loop. Lines 12 and 13 sit inside it, and the indent
is what puts them there. Move line 9 inside the loop and the counter starts
over on every bake, so the answer is never more than 1.

The four `planned and made` lines print before the answer. The function had
to walk the whole list before it had a count to give back.

## 7B. Count the misses, then work out a rate

A bake can miss the plan in two ways. It can make fewer rolls than planned,
or it can make more. Those are two different problems in a kitchen, so you
count them apart.

Three counters do that. `if`, `elif` and `else` send each bake to exactly one
of them.

A count on its own is hard to compare. Two matched bakes out of four is not
the same as two out of forty. So the last step divides the matched count by
the number of bakes. That answer is a rate.

```python
# 7B. The same four bakes. Count the two ways a bake can miss the plan, then
# divide to get the share that matched.
BAKES = [{"planned": 24, "made": 24},
         {"planned": 24, "made": 21},
         {"planned": 36, "made": 36},
         {"planned": 12, "made": 15}]

def rise_rate(bakes):
    on_target = 0
    came_up_short = 0
    ran_over = 0
    for bake in bakes:
        if bake["made"] == bake["planned"]:
            on_target = on_target + 1
        elif bake["made"] < bake["planned"]:
            came_up_short = came_up_short + 1
        else:
            ran_over = ran_over + 1
    return {"on_target": on_target, "came_up_short": came_up_short,
            "ran_over": ran_over, "rate": on_target / len(bakes)}

bake_counts = rise_rate(BAKES)
print(f"{bake_counts['on_target']} bakes of {len(BAKES)} matched the plan.")
print("Made fewer rolls than planned:", bake_counts["came_up_short"])
print("Made more rolls than planned:", bake_counts["ran_over"])
print(f"Share that matched the plan: {bake_counts['rate']}")
```

### What each line does

- Lines 3 to 6 hold the same four bakes 7A used.
- Line 8 names the function. It takes the list, under the name `bakes`.
- Lines 9, 10 and 11 start three counters at zero. One counter per group.
- Line 12 starts the loop. `bake` holds one dictionary at a time.
- Line 13 asks whether the two numbers match. Line 14 counts that bake.
- Line 15 runs only when line 13 was false. It asks whether the bake made
  fewer rolls than planned.
- Line 17 runs when both questions were false. That bake made more rolls
  than planned.
- Lines 19 and 20 give back one dictionary holding all three counts and the
  rate.
- Line 20 works the rate out. It divides the matched count by the number of
  bakes.
- Line 22 calls the function and keeps the dictionary in `bake_counts`.
- Lines 23 to 26 read the four values back out by key.

Three things to notice.

Exactly one of lines 14, 16 and 18 runs for each bake. That is what `if`,
`elif` and `else` do. So the three counts always add up to the number of
bakes, and you can check your own work by adding them.

The rate on line 20 is 2 divided by 4, which prints as 0.5. Half the bakes
matched the plan. Written that way, four bakes and forty bakes can be
compared.

Lines 23 and 26 are f-strings. The `f` in front of the quote is what makes
the braces work. The key inside the braces is in single quotes, because the
sentence around it already uses double quotes.
